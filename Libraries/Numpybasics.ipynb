Stacking


```python
#horizontal stacking :join 2/more numpy arrays of same dim
'''ex : if 2x2 & 2x2
hor 2x4 and vert 4x2'''
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)

a4
```

```python
np.hstack((a4,a5))
```

```python
#vertical stacking
#UC for stacking: diff sources of data
np.vstack((a4,a5))
```
