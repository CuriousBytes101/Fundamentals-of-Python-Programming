```python
#USE OF CONTINUE

name = "mariya mennen"   

count = 0

for char in name:
    if char != 'm':
        continue
    else:                   # char = m then count 
        count = count + 1

print('Total number of m is :',count)

#  count = 0
# m -> count = 1
# a,r,i,y,a,space != 'm'  =>just continue
# m ->  count = 1 + 1 = 2 
```

```python
#USE OF BREAK
count = 0

for i in range(1,6):
    count = count + 1
    if count > 2:
        break 
    else:
        print(i)
        
print("Done")


# count = 0 
# count = 1  print-> 1
# count = 2 print -> 2
# count 3   3>2 -> break
  
```

```python
#IMP(while loop purpose) 
# Count number of digits

count = 0

num = int(input("Enter a number:"))

while num > 0:
    num = num // 10    # Remove/chop the last digit
    count += 1         # Count the removed digit

print("Number of digits:", count)

# example: num = 1234
# 1234 // 10 = 123   last digit 4 removed -> count = 1
# 123  // 10 = 12    last digit 3 removed -> count = 2
# 12   // 10 = 1     last digit 2 removed -> count = 3
# 1    // 10 = 0    last digit 1 removed -> count = 4
# num becomes 0 -> loop stops
# Therefore, number of digits = 4

# //10 is used because it removes/chops the last digit of the number
# count += 1 counts each removed digit

# //10 removes/chops the last digit
# /10 gives a decimal value
```

```python
# Reverse a number

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10              # Get last digit
    reverse = reverse * 10 + digit # Add digit to reverse
    num = num // 10               # chop the last digit

print("Reversed no:", reverse)

# Dry run: num = 1234
# Step 1: digit = 1234 % 10 = 4
#         reverse = 0 * 10 + 4 = 4
#         num = 1234 // 10 = 123
#
# Step 2: digit = 123 % 10 = 3
#         reverse = 4 * 10 + 3 = 43
#         num = 123 // 10 = 12
#
# Step 3: digit = 12 % 10 = 2
#         reverse = 43 * 10 + 2 = 432
#         num = 12 // 10 = 1
#
# Step 4: digit = 1 % 10 = 1
#         reverse = 432 * 10 + 1 = 4321
#         num = 1 // 10 = 0
#
# num = 0 → loop ends
# Output: Reversed no: 4321
```
