```python
# Student 

class Student:
    def __init__(self,marks):
        self._marks = marks

    def update_marks(self,new_marks):
        
        if  0 <= new_marks <=100:     # new._marks >= 0 and new._marks <= 100:
            self._marks = new_marks
            print("Marks Updated")
        else:
            print("Invalid mark")

    def display_marks(self):
        print("Student Marks: ",self._marks)


s1 = Student(75)
s1.display_marks()

s1.update_marks(90)
s1.display_marks()

s1.update_marks(110)  
```

```python
#TASK 2 :
#create base class of bank acc hide and control internal data :acc no,acc_holdername & balance
#in public method withdraw,deposit,balance,etc

#same for inheritance ex :savings & current acc is subclass of main acc

#method() from saving/current resuse in currrent acc interest formula
```

```python

# Parent Class

class BankAccount:

    def __init__(self, acc_no, acc_holder, balance):

        self.__acc_no = acc_no
        self.__acc_holder = acc_holder
        self.__balance = balance

    # deposit money
    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount
            print("Amount Deposited")
        else:
            print("Invalid Amount")

    # withdraw money
    def withdraw(self, amount):

        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Amount Withdrawn")
        else:
            print("Invalid Amount")

    # display account details
    def display_details(self):

        print("Account Number:", self.__acc_no)
        print("Account Holder:", self.__acc_holder)
        print("Balance:", self.__balance)


#child class 1
class SavingsAccount(BankAccount):

    def __init__(self, acc_no, acc_holder, balance, interest):

        # Reusing parent constructor
        super().__init__(acc_no, acc_holder, balance)

        self.interest = interest

    # Calculate interest
    def calculate_interest(self):

        interest_amount = 10000 * self.interest / 100
        print("Interest:", interest_amount)


#child class 2
class CurrentAccount(BankAccount):

    def __init__(self, acc_no, acc_holder, balance):

        super().__init__(acc_no, acc_holder, balance)

    # Method Overriding
    def withdraw(self, amount):

        print("Current Account Withdrawal")

        # Reusing parent withdraw method
        super().withdraw(amount)


# Creating Savings Account Object

s1 = SavingsAccount(101, "Saniya", 10000, 5)

s1.display_details()
s1.deposit(2000)
s1.withdraw(1000)
s1.calculate_interest()

print("------------------")

# Creating Current Account Object

c1 = CurrentAccount(102, "Rahul", 15000)

c1.display_details()
c1.withdraw(5000)
c1.display_details()
```

**#Abstraction**

```python
from abc import ABC,abstractmethod 

class ATM(ABC):
    @abstractmethod
    def withdraw(self,amount):
        pass

class BankATM(ATM):
    def withdraw(self,amount):
        print("withdrawn amount:",amount)
        print("Please collect your cash")
        
atm = BankATM()
atm.withdraw(5000)
```

```python
#shape & area

from abc import ABC ,abstractmethod
class Shape(ABC):
    @abstractmethod
    def cal_area(self):
        pass

class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def cal_area(self):
        area = self.length * self.breadth
        print("Rectangle area:",area)

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius 

    def cal_area(self):
        area = 3.14 * self.radius * self.radius   #or (self.radius)**2
        print("Circle area: ",area)

r1 = Rectangle(10,5)
r1.cal_area()

c1 = Circle(4)
c1.cal_area()
     
```

```python
# 2) second largest element from list

from functools import reduce

lst = [2,5,6,7,8,15]

largest = reduce(lambda a,b : a if a > b else b,lst)

second = reduce(lambda a,b: a if a > b else b ,
                filter(lambda x:x<largest,lst))


print(second)
```
