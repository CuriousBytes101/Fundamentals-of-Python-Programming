```python
#read
file = open("demo.txt","r")
b = file.read()
print(b)
file.close()
```

```python
#readline: read single line 
file = open("demo.txt","r")
a = file.readline()
print(a)
file.close()
```

```python
#readlines : read all lines at once
file = open("demo.txt","r")
c = file.readlines()
print(c)
file.close()
```

```python
#write

file = open("demo.txt","w")
file.write("I love programming\n")
file.write("We are in write mode\n")

file.close()

#output : overwrites previous texts
```

```python
#append
#to maintain our earlier text as well

file = open("demo.txt","a")

file.write("To add more things we use append\n")
file.write("File ends here!!\n")
file.close()
```
