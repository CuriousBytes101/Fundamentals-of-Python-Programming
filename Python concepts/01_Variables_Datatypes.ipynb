**DATATYPES**

```python
'''
Basic Datatypes: 
1)str:
2)int
3)float
4)complex '''


# 1. STRING DATATYPE

text = "Hello I'm learning Python"

print("-------STRINGS--------\n")

print("Original: " ,text)
print("Length: ",len(text))
print("Uppercase:",text.upper())
print("Lowercase:",text.lower())
print("Capitalize:",text.capitalize())  #makes the first char uppercase and remaining char lowercase.
print("title:",text.title())           #first letter of each word uppercase
print("Strip:",text.strip())           #removes spaces from beginning/end

print("Replace:" ,text.replace("Hello","Hola"))

print("Find Python:", text.find("Python")) #returns index of substring
print("Find Java:", text.find("Java") )    # returns -1 if not found

print("Count of a:",text.count("a"))

# for new line \n

print("Start with Hello(T/F) ? ",text.strip().startswith("Hello"))   # returns boolean:true
print("Ends with Programming(T/F) ? ", text.strip().endswith("Programming"))

print("Split:",text.split())   # split the sentence into single word
```

```python
# 2. INTEGER DATATYPE

a = 25 
b = -10

print("-------INTEGER--------\n")

print("a =" ,a)
print("b =" ,b)

print("Absolute value of b:",abs(b))   #abs : negative to positive
print("Power:", pow(a,2))
print("Division and remainder:", divmod(16,5))   # quotient and remainder

print("Binary:", bin(10))
print("Octal:",oct(10))
print("Hexadecimal:",hex(10))
print("Maximum:" ,max(a,b))
print("Minimum:" ,min(a,b))

#Mathematical operation
print("Mathematical Operations\n")

print("Addition:" ,a + 10)
print("Substraction:" ,a - 10)
print("Multiplication:" ,a * 10)
print("Division:" ,a / 10)     #quotient
print("Floor Div:" ,a // 10)  #quotient without decimal
print("Modulus:" ,a % 10)    #remainder
print("Power:" ,a ** 2)
```

```python
# 3. FLOAT DATATYPE

price = 125.6789

print("-------FLOAT--------\n")
print("Price:",price)
print("Rounded value:", round(price,2))
print("Absolute value:", abs(-25.50))
print("Integer conv:",int(price))
print("Float conv:", float("50"))
print("Max:", max(10.5,20.2,15.8))
print("Min:", min(10.5,20.2,15.8))
print("Sum:", sum( [10.5,20.2,15.8] )) #list for sum
```

```python
# 4. COMPLEX DATATYPE
z = 3 + 4j
print("-------COMPLEX--------\n")

print("Complex no:",z)

print("Real no:", z.real)
print("Imag no:", z.imag)

print("Conjugate:", z.conjugate())
print("Magnitude:", abs(z))    # 

#Complex number operations
c1 = 3 + 4j
c2 = 2 + 5j

print("Complex Operations: \n")
print("Addition:" ,c1 + c2)
print("Substraction:" ,c1 - c2)
print("Multiplication:" ,c1 * c2)
print("Division:" ,c1 / c2 ) 
```
