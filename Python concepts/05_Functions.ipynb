```python
# Prime Number Check 

def check_prime(n):


    if n < 2:
        return False



    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False  
    return True


print(check_prime(15))  
print(check_prime(7))  
print(check_prime(6))  
```

```python
#count vowels

def count_vowels(text):
    count = 0

    for ch in text.lower():   #case-insensitivity 
        if ch in "aeiou" :     #or use ('a','e','i','o','u')
            count += 1
    return count


print(count_vowels('Saniya'))
print(count_vowels('Abby'))
print(count_vowels('It Is Raining In Pune'))
```

```python
#palindrome no.

def palindrome(n):
    if n < 0:
        return False

    rev = 0
    original = n
        
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10
        
    return original == rev
        

print(palindrome(121))
print(palindrome(14141))
print(palindrome(14231)) #F
print(palindrome(11))
```

```python
#password strength checker

def check_password(password):
    if len(password) < 8:
        return "Weak password : too short"
    has_digit = any(ch.isdigit() for ch in password)
    has_upper = any(ch.isupper() for ch in password)
    has_lower = any(ch.islower() for ch in password)

    if has_digit and has_upper and has_lower:
        return "Strong password"
    return "Moderate:Try adding uppercase,lowercase and digit" #if any 1 condition is false
  
print(check_password("S@niya456"))
print(check_password("saniyagodikat"))
print(check_password("456sango"))
```

```python
#most frequent character

def most_freq_char(text):
    frequency = {}
    for ch in text.lower():
        if ch != " ":
            frequency[ch] = frequency.get(ch,0) + 1

    if not frequency:
        return None
    return max(frequency,key=frequency.get)
            
print(most_freq_char("Hello World"))            
print(most_freq_char("banana"))     
    
```

```python
#email masking 

def mask_email(email):
    name,domain = email.split("@",1)
    if len(name) <= 2:
        hidden = name[0] + "*"
    else:
        hidden = name[:2] + "*" * (len(name) - 2)
    return hidden + "@" + domain

print(mask_email("saniya123@gmail.com"))
print(mask_email("rafatabgodikat@gmail.com"))

# Dry Run:
# Input: saniya123@gmail.com
# name = "saniya123", domain = "gmail.com"
# len(name) = 9, so hidden = name[:2] + "*" * (9 - 2)
# hidden = "sa" + "*******" = "sa*******"
# Output: sa*******@gmail.com
    
```

```python
#Find missing number for continuos series

def find(numbers,n):
    expected = n * (n + 1) // 2    # // round down
    actual = sum(numbers) 
    return expected - actual

print(find([1,2,3,5,6],6)) 
print(find([1,2,4],4))
print(find([1,2,3,4,5],5))  # 0: no missing no
print(find([2,3,4,5],5))  # characters should begin with no. 1  
```

```python
#target number pair

def pair(numbers,target):
    pairs = []
    
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                pairs.append((numbers[i],numbers[j]))
    return pairs


print(pair([1,2,3,4,5,6,7],8))
print(pair([11,2,3,5,6,16,1,1],4))
        
```

```python
#classify students as pass or fail

def class_student(marks):
    result = {"Pass": [], "Fail":[]}

    for name,score in marks.items():
        if score >= 35:
            result["Pass"].append(name)
        else:
            result["Fail"].append(name)
    return result
            
student = {"Saniya": 17,"Siddhi":95,"Rafat":81,"Sattar":99}
print(class_student(student))

# print(class_student(40))
```

```python
#shopping bill calculator

def shop_bill(cart,discount = 0):
    total = sum(cart.values())
    discount_amt = total * discount/100
    final_bill = total - discount_amt

    return {
        "Subtotal": total,
        "Discount": discount_amt,
        "Final Bill" : final_bill
    }


cart = {"chocolates":100,"Snacks":100,"medicines":100,"bag": 100}

print(shop_bill(cart,10))
```

```python
# ATM withdrawal

def withdraw(balance,amount):
    if amount <= 0:
        return "Enter a valid amount"
    if amount % 100 != 0:
        return "Enter amount in multiple of 100"
    if amount > balance:
        return "Insufficient balance"
    return balance - amount

print(withdraw(8000,1500))    
```

```python
def char_freq(text):
    frequency = {}

    for ch in text:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1
    return frequency

print(char_freq("banana"))
print(char_freq("apple"))            
```

```python
#Remove duplicates

def remove_dup(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)
    return result 
    
print(remove_dup([1,2,2,3,3,4,5,1,8]) )
```

```python
#even no - using list

def even_num(numbers):
    even = []

    for num in numbers:
        if num % 2 == 0:
            even.append(num)
    return even

print(even_num([1,2,3,4,5,6,8,10,14,15]))
```

```python
#factorial

def factorial(num):
    if num < 0 :
        return "Invalid number"     
    result = 1   
    for i in range(1,num + 1):
         result *= i
    return result 

print(factorial(5))
print(factorial(6))
print(factorial(-2))     
print(factorial(1))
print(factorial(0))
```

```python
#common elements

def common_elements(list1,list2):
    common = []

    for item in list1:
        if item in list2 and item not in common:
            common.append(item)
    return common

print(common_elements([1,3,4,6],[3,4,7,9]))  
print(common_elements([1,5,7,6,10],[3,4,7,9,6,5,20]))  
```

```python
#second largest unique no.

def sec_large(numbers):
    unique = list(set(numbers))
    unique.sort()

    if len(unique) < 2:
        return "Not enough numbers"
    return unique[-2]

print(sec_large([10,2,40,12,14,25,29,32,90]))
```

```python
#Even odd separator

def sep_num(numbers):
    even = []
    odd = []

    for num in numbers:
        if num % 2 == 0:
            even.append(num)
        else:
            odd.append(num)
            
    return even,odd

e, o = sep_num([1,2,4,5,7,6,8,11,10])
print("Even: ",e)
print("Odd: ",o)
    
```

```python
#student performance analyzer

def analyze_marks(marks):
    if not marks:
        return None
    average = sum(marks)/len(marks)

    if average >= 85:
        grade = 'Best Performance'
    elif average >= 75:
        grade = 'Good Performance'
    elif average >= 50:
        grade = 'Average Performance'
    else:
        grade = 'Need Improvement'
    return {
    "Marks": marks,
    "Average": average,
    "Grade": grade
    }
        
        

result = analyze_marks([30,45,60])    
print(result)
 
    
```
