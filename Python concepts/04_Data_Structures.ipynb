*1.LIST OPERATIONS*

```python
#1.INDEXING OR ACCESSING: 0 to n

fruits = ["Apple","Banana","Kiwi","Dragonfruit","Watermelon","Orange","Mango","Grapes"]

print(fruits)

print("------------------------------------")
print(fruits[0])
print(fruits[-0])

print(fruits[3])
print(fruits[-1])
print(fruits[-4])

#print(fruits[10])  IndexError: list index out of range
```

```python
#2.SLICING

a = [10,20,30,40,50,60]
print(a[0:3])  # 0 to 3(exclude)
print(a[:4])  # start to 4(exclude)
print(a[4:])  # 4 to end
```

```python
#3.APPEND : add new item at end
a.append(100)
a.append(110)
print(a)
```

```python
#4.INSERT: at specific position
a.insert(5,70)
print(a)
```

```python
#5.REMOVE: removes an item
a.remove(40)
print(a)
```

```python
#6.POP: removes items using index
a.pop(2)   #pop element of index 2
print(a)
```

```python
#7.EXTEND: add multiple elements

a = [1,2]
b = [3,4,5]

a.extend(b)

print(a)
```

```python
#SORT: arrangement in ascending order
d = [80,30,15,10,90,1,1,12,99]

d.sort()
print(d)

# REVERSE:descending order
d.reverse()
print(d)
```

```python
#COUNT
 
print(d.count(30))
#or

count = d.count(1)
print("No of 1:",count)

#INDEX
d = [80,30,15,10,90,1,1,12,99]
print(d.index(90))
```

```python
#join : using + operator
#repeat : using * operator
f = [1,2]
z = [3,4]

print(f + z)
print(f * 5)
```

*2.TUPLE OPERATIONS*

```python
#1.ACCESSING & INDEXING

colors = ('Red','Yellow','Blue','Pink','Green','Cyan')

print(colors[2])
print(colors[0])

print(colors[::])
print(colors[1:5]) #exclude 5
print(colors[2::])
```

*SET OPERATIONS*

```python
#1.ADD:can add single element,duplicate value ignore

s = {25,10,15,20,5}
s.add(40)   # insert at any random position:unordered(no index)
print(s)
```

```python
#2.UPDATE:add multiple elements[] in set

s.update([50,55,65])  # []
print(s)
```

```python
#3.REMOVE & DISCARD(same) :both removes element
s.remove(15)
s.discard(100)  # no error when deleting missing element

#diff:
#discard : no error when element is not there

print(s)
```

```python
s.remove(100) # will give error coz 100 doesnt exist
```

```python
s.discard(100)
print(s)
```

```python
#4.MATHEMATICAL SET OPERATIONS

# 4.1 UNION OF SETS
set1 = {1,2,3,4}
set2 = {3,4,5,6}
set3 = {1,2}
#method 1:
print(set1.union(set2))

#method 2:
print(set1|set2)
print("--------------------------------------")

# 4.2 INTERSECTION OF SETS

#method 1:
print(set1.intersection(set2))

#method 2:
print(set1 & set2)
print("--------------------------------------")
# 4.3 SUBSET & SUPERSET

print(set3.issubset(set1))
print(set3.issubset(set2))

print(set1.issuperset(set3))

```

```python
#find even no. in list
numbers = [3,4,5,6,1,10]

for num in numbers:
    if(num%2 == 0):
        print(f"Even no:{num}")
    
print("---------------------------------------------------")


#find even and odd
for i in numbers:
    if(i % 2 == 0):
        print(f"Even no:{i}")
    else:
        print(f"Odd no:{i}")
    
```

```python
#Sum of no in lists
numbers = [1,2,3,4,5]

total = 0
i = 0

while i <= len(numbers):
    total = total + i
    i+=1
    
print(total)
```
