```python
# IF STATEMENT
age = 20
if age >=18:
    print("Eligible to vote")
```

```python
#IF-ELSE STATEMENT

#ex 1
age = 16

if age >=18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")

#ex 2
salary = 35000
exp = 3

if salary > 30000 and exp >= 2:
    print("Employee is eligible for promotions")
else:
    print("Employee is not eligible for promotions")
```

```python
exp = 2

if not exp == 0:
    print("Employee has experience")
else:
    print("Fresher")
```

```python
#IF- ELIF-ELSE STATEMENT

marks = 80

# ex
if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
else:
    print("Grade C")
    
```

```python
#NESTED IF

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Sorry!Entry not allowed")
```
